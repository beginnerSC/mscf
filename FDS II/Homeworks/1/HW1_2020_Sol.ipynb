# Question 2

We only need to add the degrees of freedom to the list of parameters that are being optimized over. Note that now `pars` is of length four, with the last parameter being the degrees of freedom.

```python
def treg2(x,y):
    
    from scipy.stats import t
    from scipy.optimize import minimize
    
    def neglogliketreg(pars, x, y):
        resids = y - (x*pars[1] + pars[0])
        return (-1)*\
           sum(t.logpdf(resids/pars[2],pars[3]))+\
           len(x)*np.log(pars[2])
    
    holdcov = np.cov(x,y)
    
    beta1init = holdcov[0,1]/holdcov[0,0]
    beta0init = np.mean(y) - np.mean(x)*beta1init
    
    resids = y - (beta0init + beta1init*x)
    
    sigmainit = np.sqrt(sum(resids**2)/len(x))
    
    optout = minimize(neglogliketreg, 
       [beta0init,beta1init,sigmainit,5],
       args=(x,y), method="Nelder-Mead")
      
    return optout
```

Run some tests, for example:

```python
from scipy.stats import t
import numpy as np

x = np.array(range(1000))
y = 4 + 5*x + 100*t.rvs(10,size=len(x))

treg2(x,y).x
```

To start, it's important to look at the documentation to determine how Python parameterizes the Beta distribution. 

When we look at https://docs.scipy.org/doc/scipy-0.14.0/reference/generated/scipy.stats.beta.html, it becomes clear that Python represents the Beta in the same way as we had presented it during our Probability course.

In other words, `beta.pdf(0.4,0.8,1.2)` will evaluate the pdf of the Beta(0.8,1.2) distribution, as we defined it in Probability.

```python
def betamle(x):
    
    from scipy.stats import beta
    from scipy.optimize import minimize

    def negloglikelihood(pars,x):
        return (-1)*sum(beta.logpdf(x,pars[0],pars[1]))
    
    mleout = minimize(negloglikelihood,
       [1,1], args=(x), 
       method="Nelder-Mead")
    
    return mleout
```

```python
from scipy.stats import beta
betamle(beta.rvs(10,1,size=1000)).x
```
